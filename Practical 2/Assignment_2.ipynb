# Practical 2: Single-Layer Perceptron for Binary Classification

**Aim:** Implement a Single-Layer Perceptron for binary classification and analyze the effect of weights, bias, and activation functions.

```python
import numpy as np

# Step Activation Function
def step_function(x):
    return 1 if x >= 0 else 0

# AND Gate Dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])

# Initialize weights and bias
weights = np.zeros(2)
bias = 0

learning_rate = 0.1
epochs = 10

# Training
for epoch in range(epochs):
    print(f"Epoch {epoch+1}")
    for i in range(len(X)):
        net_input = np.dot(X[i], weights) + bias
        prediction = step_function(net_input)
        error = y[i] - prediction

        weights = weights + learning_rate * error * X[i]
        bias = bias + learning_rate * error

        print(f"Input: {X[i]}, Prediction: {prediction}, Error: {error}")
    print("-"*40)

print("\nFinal Weights:", weights)
print("Final Bias:", bias)

# Testing
print("\nTesting:")
for i in range(len(X)):
    net_input = np.dot(X[i], weights) + bias
    prediction = step_function(net_input)
    print(f"Input: {X[i]} -> Output: {prediction}")
```

## Analysis

### Effect of Weights
- Weights determine the importance of each input feature.
- Higher weights increase the influence of a feature on the output.

### Effect of Bias
- Bias shifts the decision boundary.
- It allows correct classification even when the data is not centered at the origin.

### Effect of Activation Function
- **Step Function:** Produces binary output (0 or 1), suitable for perceptrons.
- **Sigmoid:** Produces values between 0 and 1.
- **Tanh:** Produces values between -1 and 1.
- **ReLU:** Outputs max(0, x), commonly used in deep learning.

## Conclusion
The Single-Layer Perceptron correctly classifies the AND gate dataset by learning suitable weights and bias. It demonstrates binary classification for linearly separable data.
